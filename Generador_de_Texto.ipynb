Generador de Texto con GPT-2

1. Configuración del entorno y modelo
2. Text generation

```python
#1. Instalación de libreria

!pip install transformers
!pip install torch
```

```python
#Configuracion del entorno

from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

tokenizer = AutoTokenizer.from_pretrained("flax-community/papuGaPT2")
model = AutoModelForCausalLM.from_pretrained("flax-community/papuGaPT2")

generator = pipeline("text-generation", model=model, tokenizer=tokenizer)
```

```python
#2. Generador de texto

text = generator("Artificial Inteligence is", max_length=20, temperature=0.5)
print(text[0]['generated_text'])
```

```python
text = generator("Responder en español: La inteligencia artifical es", max_length=15, temperature=0.5)

print(text[0]['generated_text'])
```

```python
#Parameteros
#top_p = % probabilidad
#repetition_penalty = penalizarrepeticiones
#top_k = familia de palabras
#do_sample = improviso o no improviso


text = generator("Artificial Inteligence is",
                 max_new_tokens=20,
                 temperature=0.2,
                 top_p=0.92,
                 repetition_penalty=1.15,
                 top_k=60,
                 do_sample=True,
                 )

print(text[0]['generated_text'])
```

```python
text = generator("I am detective Sherlock Holmes, I am conducting an investigation. I have just noticed that on the desk",
                 max_new_tokens=60,
                 temperature=1.1,
                 top_p=0.92,
                 repetition_penalty=1,
                 top_k=60,
                 do_sample=False,
                 )

print(text[0]['generated_text'])
```
