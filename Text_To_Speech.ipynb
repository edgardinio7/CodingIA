```python
#3. Funciones para cambiar tono

from pydub import AudioSegment
from pydub.playback import play
from IPython.display import Audio

#Funciones simples
def change_pitch(sound, semitones=0):
    rate = int(sound.frame_rate * (2 ** (semitones/12)))
    return sound._spawn(sound.raw_data, overrides={'frame_rate': rate}).set_frame_rate(sound.frame_rate)

def change_speed(sound, speed=1.0):
    rate = int(sound.frame_rate * speed)
    return sound._spawn(sound.raw_data, overrides={'frame_rate': rate}).set_frame_rate(sound.frame_rate)

#Cargar el audio
sound = AudioSegment.from_file("hello.mp3")

#Transformación
fast = change_speed(sound, 2.5)
lower_pitch = change_pitch(sound, -3)
higher_pitch = change_pitch(sound, 8)

fast.export("rapido.mp3", format="mp3")
lower_pitch.export("bajo.mp3", format="mp3")
higher_pitch.export("alto.mp3", format="mp3")

# --- Playback----
print("Voz mas rapida:")
display(Audio("rapido.mp3"))

print("Bajos:")
display(Audio("bajo.mp3"))

print("Altos:")
display(Audio("alto.mp3"))

```
